## getting used to yahoo finance library


```python
''' downloading single company records '''
import yfinance as yf
import matplotlib.pyplot as plt

#yf.download() downloads stock data as pandas dataframe
df = yf.download(
    "SXR8.DE",
    #period="1mo",
    start = "2025-01-01",
    end = "2026-01-01",             #useful after we extract oldest transactions from XTB csv file
    interval="1d",
    auto_adjust=False,
    multi_level_index=False,
    progress=False
)

df_close = df['Close']
print(df_close)

print("last month statistics: ")
print("lowest close: ", df['Close'].min())
print("highest close: ", df['Close'].max())

```

```python
plt.figure(figsize=(12, 6))
plt.plot(df_close)
plt.title("iShares Core S&P 500 UCITS ETF USD (Acc) (SXR8.DE)")
plt.grid()
plt.show()
```
