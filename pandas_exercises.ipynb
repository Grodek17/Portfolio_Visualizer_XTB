```python
### exercise 2 ###

import pandas as pd

data = {
    "Date": [
        "2026-07-13",
        "2026-07-14",
        "2026-07-15",
        "2026-07-13",
        "2026-07-14",
        "2026-07-15",
    ],
    "Ticker": [
        "CDR.WA",
        "CDR.WA",
        "CDR.WA",
        "PKN.WA",
        "PKN.WA",
        "PKN.WA",
    ],
    "Open": [
        250.0,
        253.5,
        256.0,
        72.0,
        71.5,
        73.0,
    ],
    "Close": [
        252.4,
        255.1,
        249.8,
        71.5,
        73.2,
        74.1,
    ],
    "Volume": [
        650_000,
        720_000,
        810_000,
        1_200_000,
        980_000,
        1_350_000,
    ],
}

df = pd.DataFrame(data)
df.head(6)
```

```python
### filtering data


# show only records with closing date greater than 100
df[df["Close"] > 100]
```

```python
# show only records with cd project red tickers
df[df["Ticker"] == "PKN.WA"]
```

```python
#joining multiple statements [AND]
df[ (df["Ticker"] == "CDR.WA") & (df["Close"] > 250) ]
```

```python
#multiple conditionals [OR]
df[   (df["Close"] > 250) | (df["Volume"] > 1_000_000) ]
```

```python
#.loc allows filtering and choosing columns at the same time 
df.loc[df["Ticker"] == "CDR.WA", ["Date", "Close"]]
```
